```python
base_url = "https://books.toscrape.com/catalogue/page-{}.html"

books = []

for page in range(1, 51):
    url = base_url.format(page)

    response = req.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    articles = soup.find_all("article", class_="product_pod")

    for article in articles:
        books.append({
            "Title": article.h3.a["title"],
            "Price": article.find("p", class_="price_color").text,
            "Rating": article.find("p", class_="star-rating")["class"][1],
            "Availability": article.find("p", class_="instock availability").text.strip()
        })

print(books[0])   # First 5 books
print("Total:", len(books))
```

```python
base_url = "https://books.toscrape.com/catalogue/page-{}.html"

books = []

for page in range(1, 51):
    soup = BeautifulSoup(req.get(base_url.format(page)).text, "html.parser")

    articles = soup.find_all("article", class_="product_pod")

    for article in articles:
        # Relative link
        href = article.h3.a["href"]

        # Full book URL
        book_url = "https://books.toscrape.com/catalogue/" + href.replace("../", "")

        book_soup = BeautifulSoup(req.get(book_url).text, "html.parser")

        category = book_soup.find("ul", class_="breadcrumb").find_all("li")[2].text.strip()

        books.append({
            "Title": article.h3.a["title"],
            "Category": category
        })

print(books[:5])
```

```python
# URL template for all 50 pages
# {} will be replaced by page number (1, 2, 3, ...)
base_url = "https://books.toscrape.com/catalogue/page-{}.html"

# Empty list to store data of all books
books = []

# Loop through all 50 pages
for page in range(1, 51):

    # Create URL for current page
    url = base_url.format(page)

    # Send HTTP GET request to the page
    response = req.get(url)

    # Parse HTML content using BeautifulSoup
    soup = BeautifulSoup(response.text, "html.parser")

    # Find all book containers (<article class="product_pod">)
    articles = soup.find_all("article", class_="product_pod")

    # Loop through each book on the current page
    for article in articles:

        # Extract relative URL of the current book
        href = article.h3.a["href"]

        # Convert relative URL to complete URL
        book_url = "https://books.toscrape.com/catalogue/" + href.replace("../", "")

        # Send request to the individual book page
        book_response = req.get(book_url)

        # Parse the HTML of the individual book page
        book_soup = BeautifulSoup(book_response.text, "html.parser")

        # Extract category from breadcrumb navigation
        category = (
            book_soup.find("ul", class_="breadcrumb")
                     .find_all("li")[2]
                     .text
                     .strip()
        )

        # Store all extracted information as a dictionary
        books.append({
            "Title": article.h3.a["title"],

            # Extract price
            "Price": article.find("p", class_="price_color").text,

            # Extract rating (One, Two, Three, Four, Five)
            "Rating": article.find("p", class_="star-rating")["class"][1],

            # Extract stock availability
            "Availability": article.find(
                "p",
                class_="instock availability"
            ).text.strip(),

            # Extract category from individual book page
            "Category": category
        })

# Print first 5 book records
print(books[:5])

# Print total number of books scraped
print("Total:", len(books))
```

```python
import pandas as pd

df = pd.DataFrame(books)
print(df.head())
df.shape
```
