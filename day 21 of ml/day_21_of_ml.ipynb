## 3. boxplot (numerical - categorical)

```python
# sns.boxenplot(data=titanic, x='Sex', y='Age')
sns.boxplot(data=titanic, x='Sex', y='Age', hue='Survived')
```

## 4. distplot (numerical - categorical)

```python
# sns.distplot(titanic[titanic['Survived']==0]['Age'])

sns.histplot(
    data=titanic[titanic['Survived'] == 0]['Age'],
    kde=True,
)

sns.histplot(
    data=titanic[titanic['Survived'] == 1]['Age'],
    kde=True
)
```

```python
sns.kdeplot( data=titanic[titanic['Survived'] == 0], x='Age', fill=True, label='Not Survived')

sns.kdeplot( data=titanic[titanic['Survived'] == 1], x='Age', fill=True, label='Survived')
```

```python
sns.displot(data=titanic,x="Age",hue="Survived",kind="kde",fill=True)
```

```python
sns.distplot(
    titanic[titanic["Survived"] == 0]["Age"],
    hist=False,
    label="Not Survived"
)

sns.distplot(
    titanic[titanic["Survived"] == 1]["Age"],
    hist=False,
    label="Survived"
)

plt.legend()
plt.show()
```

## 5. heatmap (categorical - categorical)

```python
sns.heatmap(pd.crosstab(titanic['Pclass'], titanic['Survived']))
```
