```python
import pickle
import numpy as np
```

```python
ohe_sex = pickle.load(open('/content/ohe_sex.pkl', 'rb'))
ohe_embarked = pickle.load(open('/content/ohe_embarked.pkl', 'rb'))
clf = pickle.load(open('/content/clf.pkl', 'rb'))
```

```python
# Assume user input
# Pclass/gender/age/SibSp/Parch/Fare/Embarked
# test_input = np.array([2, 'male', 31.0, 0, 0, 10.5, 'S'],dtype=object).reshape(1,7)
```

```python
# custom input

pclass = int(input('enter your pclass (1/2/3): '))
gender = input('enter your gender (male/female): ')
age = float(input('enter your age: '))
sibsp = int(input('enter your sibling with you(1 to 6): '))
parch = int(input('enter your parents with you: '))
fare = float(input('enter your fare:'))
embarked = input('enter your embarked(S/Q/C)')

test_input = np.array([pclass, gender, age, sibsp, parch, fare, embarked],dtype=object).reshape(1,7)
```

```python
test_input_sex = ohe_sex.transform(test_input[:, 1].reshape(1,1))
```

```python
test_input_sex
```

```python
test_input_embarked = ohe_embarked.transform(test_input[:, -1].reshape(1,1))
```

```python
test_input_embarked
```

```python
test_input_age = test_input[:, 2].reshape(1,1)
```

```python
test_input_age
```

```python
test_input_transformed = np.concatenate((test_input[:,[0,3,4,5]], test_input_age, test_input_sex, test_input_embarked), axis=1)
```

```python
test_input_transformed
```

```python
test_input_transformed.shape
```

```python
clf.predict(test_input_transformed)
```
