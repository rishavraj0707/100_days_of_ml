```python
import pandas as pd
import requests as req
```

```python
# https://api.themoviedb.org/3/movie/top_rated?api_key=b31a256a144e0a4eff0c377a376df27a&language=en-US&page=1
# api key
response = req.get('https://api.themoviedb.org/3/movie/top_rated?api_key=b31a256a144e0a4eff0c377a376df27a&language=en-US&page=1')

total_pages = response.json()['total_pages']
total_pages
```

```python
response.json()['results']
```

```python
movies = pd.DataFrame(response.json()['results'])[['id', 'title', 'release_date', 'overview', 'popularity', 'vote_average', 'vote_count']]
```

```python
df = pd.DataFrame()
```

```python
df
```

```python
Max_pages = 500
import time
start = time.time()

for i in range(1, Max_pages + 1):
  response = req.get('https://api.themoviedb.org/3/movie/top_rated?api_key=b31a256a144e0a4eff0c377a376df27a&language=en-US&page={}'.format(i))
  temp_df = pd.DataFrame(response.json()['results'])[['id', 'title', 'release_date', 'overview', 'popularity', 'vote_average', 'vote_count']]
  df = pd.concat([df, temp_df], ignore_index=True)

end = time.time()
print(f"Time taken: {end - start:.2f} seconds")
```
